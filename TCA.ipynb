```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#replacing blanks with 0 as tenure is 0 and no total charges are recorded

```python
df['TotalCharges'] = df['TotalCharges'].replace(" ", "0")
df['TotalCharges'] = df['TotalCharges'].astype("float")
```

```python
def conv(value):
    if value == 1:
        return 'yes'
    else:
        return 'no'

df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)
```

#converted 0 and 1 values of senior citizen to yes/no to make it easier to understand 

```python
df.head(25)
```

```python

ax = sns.countplot(x='Churn', data=df)
ax.bar_label(ax.containers[0])
plt.title('count of Customer by Churn')
plt.show()
```

```python
plt.figure(figsize = (4, 4))
gb = df.groupby("Churn").agg({'Churn': 'count'})
plt.pie(gb['Churn'], labels=gb.index, autopct='%1.2f%%')
plt.title("Percentage of Churned customer")
plt.show()
```

#from the given pie chart we can conclude that 26.54% of our customer have churned out # not let's explore the reason behind it 

```python
plt.figure(figsize=(5, 4))
sns.countplot(x='gender', data=df, hue='Churn')
plt.show()
```

```python
plt.figure(figsize=(4, 4))
ax = sns.countplot(x='SeniorCitizen', data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of Customer by SeniorCitizen")
plt.show()
```

```python
plt.figure(figsize=(5, 4))

# Calculate percentages
crosstab = pd.crosstab(df['SeniorCitizen'], df['Churn'], normalize='index') * 100

# Plot stacked bar
ax = crosstab.plot(kind='bar', stacked=True, colormap='Set2', ax=plt.gca())

# Add percentage labels on each segment
for bars in ax.containers:
    ax.bar_label(bars, 
                 labels=[f'{v:.1f}%' for v in bars.datavalues], 
                 label_type='center', 
                 fontsize=10)

plt.title('Churned by SeniorCitizen')
plt.xlabel('SeniorCitizen')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Churn', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
#comparative a greater percentage of people in senior citizen category have churned
```

```python
cols = ['PhoneService', 'MultipleLines', 'InternetService',
        'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']

fig, axes = plt.subplots(3, 3, figsize=(14, 10))
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.countplot(x=col, data=df, ax=axes[i], hue=df['Churn'], legend=False, palette='Set2')
    axes[i].set_title(f'Count of {col}')
    axes[i].set_xlabel('')
    for container in axes[i].containers:
        axes[i].bar_label(container)

plt.tight_layout()
plt.show()
```

```python
#The majority of customers have Phone Service (~4662) but only about half have Multiple Lines. For Internet Service, DSL and Fiber Optic are the dominant types with very few having no internet. Across add-on services like Online Security, Tech Support, Online Backup, and Device Protection, most customers have opted out ("No"), suggesting low adoption of premium features
```
